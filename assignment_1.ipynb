2. String Slicing and Indexing:

```python
print(msg)
print("Length of the string : ", len(msg))
print("First character of the string : ", msg[0]) # Print the first character of the string.
print("Last character of the string : ", msg[-1]) # Print the last character of the string.
print("First 5 characters of the string : ",msg[0:6:]) # Print the first 5 characters of the string.
print("Last 11 characters of the string : ",msg[-11::]) # Print the last 11 characters of the string.
print("Reverse of the string: ",msg[::-1])
print(msg[25:31]) # Use slicing and print the word “Python” from the existing string.
```

3. String Methods:

```python
strM = "Python beginner tutorial"
print(strM)
print("Uppercase : ",strM.upper()) # Convert the sentence to uppercase
print("Lowercase : ",strM.lower()) # Convert the sentence to lowercase
print("Capitalize then return to original form: ",strM.capitalize()) # Capitalize the sentence
print("Occurrences of character 't' : ",strM.count('t')) # occurrences of character ‘t’
print("New String : ",strM.replace("Python"," Machine Learning")) # Replace "Python" with "Machine Learning"
```

Tuples (Creation, Modifi cation and Access)

```python
tup1= (10,20,30)
tup2= (40,50,60)
print("Tuple 1: ",tup1)
print("Tuple 2: ",tup2)
```

```python
t_combine= tup1 + tup2 # Concatenate the two tuples and store it in “t_combine”
print(t_combine)

print(t_combine *3 )# Repeat the elements of “t_combine” 3 times

print(t_combine[2]) # Access the 3rd element from “t_combine”

print(t_combine[:3]) # Access the first three elements from “t_combine”
print(t_combine[-3:]) # Access the last three elements from “t_combine”
```
